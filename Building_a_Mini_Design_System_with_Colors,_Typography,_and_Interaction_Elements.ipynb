1.     Define a color system that ensures consistency and accessibility?

                 Color System for Consistency and Accessibility

>      A color system is a structured set of colors, rules, and design tokens
       used throughout a digital product to maintain a consistent visual
       appearance while ensuring that text and interface elements are
       accessibile to all users.

                                Key Elements

       >   Primary Colors- Used for important actions, branding,and main
           interface elements.
       >   Secondary Colors- Support the primary palette and create visual
           variety.
       >   Neutral Colors- Include white, black, and gray for backgrounds,
           text, borders, and surfaces.
       >   Semantic Colors- Use consistent colors for meanings such as
           success, warning, error, and information.
       >   Contrast- Ensure sufficient contrast between text/ icons and their
           backgrounds, following accessibility guidelines such as WCAG.
       >   Color Tokens- Define reusable names such as primary-500, text-
           primary, and error- 500 so colors remain consistent across screens.
       >   Don't Rely Only On Color- Use icons, labels, patterns, or text
           along with color to communicate inportant information, helping
           users with color- vision deficiencies.

b.      Apply WCAG - compliant contrast ratios to ensure readability?

                Apply WCAG- Complaint Contrast Ratios

>       Ensure that the UI follows WCAG ( Web Content Accessibility
        Guidelines) contrast requirements so text and important interface
        elements remain readable for users with visual impairements.

     >  Normal Text- Minimum contrast ratio of 4:5:1 against its background.
     >  Large Text-  Minimum 3:1 contrast ratio. Large text generally means
        18pt+ or 14pt+ bold.
     >  UI Components & Graphics- Important controls, icons, borders, and
        visual indicators should generally maintain at least 3:1 contrast against adjacent colors.
     >  Don't Rely on Color Alone- Use labels, icons, patterns, or text to
        communicate important information.
     >  Test the color Palette- Use tools such as Figma's Stark Plugin,
        WebAIM Contrast Checker, or accessible color- checking tools to verify every text/background combination.

2.     Establish a typography hierarchy to maintain readability and structure?

                     Establish a Typography Hierarchy

>      Typography hierarchy means arranging font sizes, weights, and styles in
       a clear order so users can easily understand the importance of different pieces of content.
       A basic hierarchy can be:-
    >  H1- Main Heading: 32-40Px, bold
    >  H2- Section Heading 24-32 px, semibold
    >  H3- Subheading 20-24 px,semibold
    >  Body Text:- 16px, regular
    >  Small/ Supporting Text:- 12-14 px, regular

       Key Points-

    >  Use larger fonts for important headings.
    >  Use bold or semibold weights to highlight headings.
    >  Maintain consistent line spacing and margins.
    >  Use a limited number of font styles to avoid visual clutter.
    >  Ensure sufficient contrast between text and background.
    >  Keep body text around 16px or larger for comfortable reading.

3.   Develop iconography guidelines to maintain a consistent visual
     language?

           Iconography Guidelines for a Consistent Visual Language

>    To maintain a consistent visual language, define clear rules for how
     icons are designed and used across the product:-
   1. Choose one icon Style:- Use a consistent style such as outline, filled,
      or two-one. Avoid mixing different styles.
   2. Maintain consistent size:- Define standard sizes such as 16px, 20px,
      and 24px and use them consistently.
   3. Use a consistent stroke weight- Keep the same stroke thickness, for
      example 2px, across icons.
   4. Keep proportions Consistent:- Icons should follow a common grid and
      have similar visual weight and spacing.
   5. Use a standard icon grid:- Design icons on a fixed grid, such as
      24 * 24 px grid, to ensure alignment.
   6. Apply consistent corner styles:- Use similar rounded or sharp corners
      throughout the icon set.
   7. Maintain Proper Spacing:- Provide adequate padding around icons and
      keep icon- to- text spacing consistent.
   8. Use Meaningful Icons:- Icons should be simple, recognizable, and
      clearly communicate their function.
   9. Follow Accessibility Guidelines:- Ensure sufficient contrast and do not
      rely on icons alone to communicate important information.
   10. Define usage rules:- Document where icons should be used, including
       navigation, buttons, alerts, actions, and status indicators.
   11. Create Reusable Components:- Store icons as reusable components
       in Figma so updates remain consistent throughout the design system.
   12. Avoid Unnecessary Decoration:- Keep icons simple and functional
       rather than adding excessive visual details.

b.    Define icon size, spacing, and alignment rules for consistency?

                   Define icon size, spacing & Alignment Rules

>     To maintain a consistent and professional icon system, establish
      clear rules for size, spacing and alignment:-

>     Icon Sizes- Use standard sizes such as 16px, 20px, 24px, and 32px
      depending on the component and context.
>     Spacing- Maintain consistent spacing between icons and text,
      typically 8px or 12px.
>     Alignment- Align icons vertically and horizontally with nearby text
      and UI elements using a consistent grid.
>     Padding- Give icons sufficient clickable area, ideally 44 * 44px
      or larger for touch interfaces.
>     Consistency- Use the same size and alignment rules across buttons,
      navigation bars, cards, forms, and other components.
>     Optical Alignment- Adjust icons slightly when necessary so
      they appear visually centered, even if their bounding boxes are mathematically centered.

5.  Implement microinteractions and feedback mechanisms for usability?

      Implement Microinteractions and Feedback Mechanisms For Usability

>   Microinteractions are small animations or visual responses that tell
    users what is happening when they interact with a UI. They make the
    interface feel responsive, clear, and easy to use.

  1. Button Interactions-

     Add different states to buttons-
  >  Hover- Button slightly changes appearance when the cursor is over it.
  >  Press- Button changes briefly when clicked.
  >  Disabled- Button appears faded and cannot be clicked.
  >  Success- Show a checkmark or confirmation after an action.

  2.  Form Validation-

      Give immediate feedback when users enter information.
   >  Correct Input n show or success message.
   >  Incorrect input n highlight the field and explain the problem.
   >  Empty required field n display " This field is required".

   3. Loading Feedback-

      When an action takes time, show-
    > Loading Spinner
    > Progress Bar
    > Skeleton Screen
      This prevents users from thinking the application has stopped working.

    4. Notifications-

       Use toast messages, banners, or alerts to confirm important actions.

    5. Hover Tooltips-

       Use tooltips to explain unfamiliar icons or controls.

     6. Subtitle Animations-

        Use short and simple animations for-
     >  Opening/ Closing Menus
     >  Switching Tabs
     >  Showing Modals
     >  Adding items to a cart
     >  Completing a task
        Avoid excessive animation because it can distract users.
